```python

%%writefile merge_sort.c

#include <stdio.h>
#include <stdlib.h>
#include <time.h>

long long merge(int* vetor, int comeco, int meio, int fim) {
  long long count = 0;
  #define QTD_EXECUT(x) (count += (x))
  #define TESTE_WHILE(p1, meio, p2, fim) \
    (QTD_EXECUT(1), (p1 <= meio) ? (QTD_EXECUT(1), QTD_EXECUT(1), (p2 <= fim)) : (QTD_EXECUT(1), 0))
  QTD_EXECUT(3);



    int posicao1 = comeco;
    QTD_EXECUT(1);
    int posicao2 = meio + 1;
    QTD_EXECUT(1);
    int posicaoAux = 0;
    QTD_EXECUT(1);

    int tam = fim - comeco + 1;
    QTD_EXECUT(1);

    int* vetAux;
    QTD_EXECUT(1);

    vetAux = (int*)malloc(tam * sizeof(int));
    QTD_EXECUT(1);



    while (TESTE_WHILE(posicao1, meio, posicao2, fim)) {

        QTD_EXECUT(1);
        if (vetor[posicao1] < vetor[posicao2]) {

            vetAux[posicaoAux] = vetor[posicao1];
            QTD_EXECUT(1);
            posicao1++;
            QTD_EXECUT(1);

        } else {

            vetAux[posicaoAux] = vetor[posicao2];
            QTD_EXECUT(1);
            posicao2++;
            QTD_EXECUT(1);
        }

        posicaoAux++;
        QTD_EXECUT(1);
    }

    // Elementos restantes da primeira metade
    QTD_EXECUT(1); // primeira verificação
    while (posicao1 <= meio) {

        vetAux[posicaoAux] = vetor[posicao1];
        QTD_EXECUT(1);

        posicaoAux++;
        QTD_EXECUT(1);
        posicao1++;
        QTD_EXECUT(1);

         QTD_EXECUT(1); // próxima verificação do while
    }

    // Elementos restantes da segunda metade
    QTD_EXECUT(1); // primeira verificação do while
    while (posicao2 <= fim) {

        vetAux[posicaoAux] = vetor[posicao2];
        QTD_EXECUT(1);

        posicaoAux++;
        QTD_EXECUT(1);
        posicao2++;
        QTD_EXECUT(1);

        QTD_EXECUT(1); // próxima verificação do while
    }

    // Volta para o vetor original
      for (posicaoAux = comeco; posicaoAux <= fim; posicaoAux++) {
            QTD_EXECUT(1); // posicaoAux = comeco
            QTD_EXECUT(1); // posicaoAux <= fim

             vetor[posicaoAux] =
                 vetAux[posicaoAux - comeco];

             QTD_EXECUT(1); // atribuição

             QTD_EXECUT(1); // posicaoAux++
     }

    // printf("\n");

    free(vetAux);
    QTD_EXECUT(1);

    #undef QTD_EXECUT
    #undef TESTE_WHILE

    return count;

}


long long mergeSort(int* vetor, int comeco, int fim) {
  long long count = 0;
  #define QTD_EXECUT(x) (count += (x))
  QTD_EXECUT(2);

    QTD_EXECUT(1);
    if (comeco < fim) {

        int meio = (fim + comeco) / 2;
        QTD_EXECUT(1);

        // Removemos o printf intermediário para uma saída mais limpa
        // for (int i = comeco; i <= meio; i++) {
        //     printf("%d ");
        // }

        // printf("\n");
        QTD_EXECUT(1);
        count +=  mergeSort(vetor, comeco, meio);



        // Removemos o printf intermediário para uma saída mais limpa
        // for (int i = meio + 1; i <= fim; i++) {
        //     printf("%d ");
        // }

        // printf("\n");
        QTD_EXECUT(1);
        count += mergeSort(vetor, meio + 1, fim);

        QTD_EXECUT(1);
        count += merge(vetor, comeco, meio, fim);


        #undef QTD_EXECUT


        return count;

    }

     return count;


}

int main() {
    int i, n, opcao;
    int tamanho_usuario = 0;
    int *vetorOriginal = NULL;
    int *vetorOrdenado = NULL;
    int QTD;
    int opcaosort;
    int algoritmoEscolhido = 0;
    int imp_done = 0;
    while (1){
      int resposta;
      printf("\n =============================================\n");
      printf("1- Escolher Tamanho\n");
      printf("2- Gerar Dados \n");
      printf("3- Ver Dados \n");
      printf("4- Escolher Algoritimo e Executar Ordenação \n");
      printf("5- Ver Resultado \n");
      printf("6- Excluir Dados \n");
      printf("7- Salvar tudo em arquivo \n");
      printf("8- Importar arquivo com dados\n");
      printf("9- Sair \n");
      printf("=============================================\n");


      char entrada[100]; // caractere que guarda ex: 'a' 'a' 'a' '\n' '\0'

      printf("Digite sua Escolha:");

      fgets(entrada,  // onde guardar o que foi digitado
            sizeof(entrada),
            stdin ); // teclado
      // Leia o que o usuário digitar e coloque dentro de entrada. entrada = "111\n"

      if (entrada[0] == '\n') { // se usuário apertou apenas enter
        printf("Opção errada.\n");
        continue;
      }

      resposta = sscanf(entrada, "%d", &opcao);

      if (resposta != 1) {
        printf("Tamanho inválido.\n");
        continue;
      }

      // isso tudo dá certo porque o teclado neste caso está no while e antes dos acontecimentos de outros scan
      switch (opcao) {
        case 1:
          if (imp_done == 1) {
                printf("Vetor já importado, vá para os métodos ordenadores.\n");
                break;
              }
            while (1){
              char entrada[100]; // caractere que guarda ex: 'a' 'a' 'a' '\n' '\0'
              printf("Digite o tamanho do vetor: ");


              fgets(entrada,  // onde guardar o que foi digitado
                    sizeof(entrada),
                    stdin ); // teclado
              // Leia o que o usuário digitar e coloque dentro de entrada. entrada = "111\n"

              if (entrada[0] == '\n') { // se usuário apertou apenas enter
                printf("Tamanho inválido. \n");
                continue;
              }

              int resultado;
              resultado = sscanf(entrada, "%d", &tamanho_usuario);

              if (resultado != 1) {
                printf("Tamanho inválido.\n");
                continue;
              }

              if (tamanho_usuario > 0) {
                printf("Tamanho escolhido: %d\n", tamanho_usuario);
                break;
              } else {
                printf("Tamanho inválido. \n");
              }
            }
            break;
        case 2:
            if (imp_done == 1) {
                printf("Vetor já importado, vá para os métodos ordenadores.\n");
                break;
              }
            if (tamanho_usuario <= 0) {
              printf("Nenhum tamanho carregado para o vetor.\n");
              break;
            }
            // Alocacao dinamica para o vetor
             vetorOriginal = (int*)malloc(tamanho_usuario * sizeof(int));
             vetorOrdenado = (int*)malloc(tamanho_usuario * sizeof(int));

            if (vetorOriginal == NULL || vetorOrdenado == NULL) {
              printf("Erro de alocacao de memoria!\n");
              return 1;
            }

            srand(time(NULL));
            // Cria números aleatórios
            for (i = 0; i < tamanho_usuario; i++) {
              n = rand();
              vetorOriginal[i] = n;
            }  // não altere isso

            for (i = 0; i < tamanho_usuario; i++) {
              vetorOrdenado[i] = vetorOriginal[i];
            }

            break;

        case 3:
            if (tamanho_usuario <= 0) {
              printf("Nenhum tamanho carregado para o vetor.\n");
              break;
            }

             if (vetorOriginal == NULL || vetorOrdenado == NULL) {
              printf("Erro de alocacao de memoria!\n");
              return 1;
            }

            printf("Vetor original:\n");
            printf("---------------------------------------------\n");

            for (i = 0; i < tamanho_usuario; i++) {
              printf("%d ",vetorOriginal[i]);
            }

            printf("\n---------------------------------------------\n");
            break;

        case 4: {
            int resultadosort;
            int sairalgoritimo = 0;

            if (tamanho_usuario <= 0) {
              printf("Nenhum tamanho carregado para o vetor.\n");
              break;
            }

             if (vetorOriginal == NULL || vetorOrdenado == NULL) {
              printf("Erro de alocacao de memoria!\n");
              return 1;
            }
            while(1){
              char entrada[100]; // caractere que guarda ex: 'a' 'a' 'a' '\n' '\0'
              printf("=============================================\n");
              printf("Escolha um metodo de ordenacao:\n");
              printf("=============================================\n");
              printf("1- Merge Sort\n");
              printf("2- Bubble Sort\n");
              printf("3- Heap Sort\n");
              printf("4- Outro método\n");
              printf("=============================================\n");

              printf("Digite sua Escolha:");


              fgets(entrada,  // onde guardar o que foi digitado
                    sizeof(entrada),
                    stdin ); // teclado
              // Leia o que o usuário digitar e coloque dentro de entrada. entrada = "111\n"

              if (entrada[0] == '\n') { // se usuário apertou apenas enter
                printf("Tamanho inválido. \n");
                continue;
              }

              int resultado;
              resultadosort= sscanf(entrada, "%d", &opcaosort);

              if (resultadosort != 1) { // scanf não leu um número inteiro
                printf("Entrada invalida!\n");
                continue;
              }
              // se for um numero inteiro positivo o resultado = 1
              if (opcaosort > 0) {
                switch (opcaosort) {
                  case 1:
                    printf("Ordenacao com MergeSort:\n");
                    printf("=============================================\n");
                    // Chamada do Merge Sort
                    QTD = mergeSort(vetorOrdenado, 0, tamanho_usuario - 1);
                    algoritmoEscolhido = 1;
                    printf("Tempo de execucao: %d\n", QTD);
                    sairalgoritimo = 1;
                    break;
                  case 2:
                    printf("Ordenacao com BubbleSort:\n");
                    algoritmoEscolhido = 1;
                    printf("=============================================\n");
                    sairalgoritimo = 1;
                    break;

                  case 3:
                    printf("Ordenacao com HeapSort:\n");
                    algoritmoEscolhido = 1;
                    printf("=============================================\n");
                    sairalgoritimo = 1;
                    break;

                  case 4:
                    printf("Ordenacao com OutroSort:\n");
                    sairalgoritimo = 1;
                    printf("=============================================\n");
                    break;

                  default:
                    printf("Opção inválida!\n");
                    continue;
            }
              } if (sairalgoritimo == 1) {
                  break;
                }
            }
          break;
        }

        case 5:
             if (tamanho_usuario <= 0) {
              printf("Nenhum tamanho carregado para o vetor.\n");
              break;
            }
             if (vetorOriginal == NULL || vetorOrdenado == NULL) {
              printf("Erro de alocacao de memoria!\n");
              return 1;
            }
             printf("\nVetor ordenado:\n");
             printf("---------------------------------------------\n");
             for (i = 0; i < tamanho_usuario; i++) {
                printf("%d ", vetorOrdenado[i]);
             }
            printf("\n---------------------------------------------\n");

            break;

        case 6:
             printf("=============================================\n");
             printf("Excluir dados...\n");
             free(vetorOriginal);
             free(vetorOrdenado);

             vetorOriginal = NULL;
             vetorOrdenado = NULL;
             printf("=============================================\n");
            break;
        case 7: {
            if (tamanho_usuario <= 0) {
              printf("Nenhum tamanho carregado para o vetor.\n");
              break;
            }
             if (vetorOriginal == NULL || vetorOrdenado == NULL) {
              printf("Erro de alocacao de memoria!\n");
              return 1;
            }
            if (algoritmoEscolhido == 0) {
              printf("Nenhum algoritmo foi executado.\n");
              break;
            }
             printf("=============================================\n");
             printf("Salvando arquivo...\n");
             FILE *arquivo;

            arquivo = fopen("resultado.txt", "w");

            fprintf(arquivo, "Vetor : ");

            for (int i = 0; i < tamanho_usuario; i++) {
              fprintf(arquivo, "%d ", vetorOriginal[i]);
            }

            fprintf(arquivo, "\n");

            fprintf(arquivo, "Vetor Ordenado: ");

            for (int i = 0; i < tamanho_usuario; i++) {
              fprintf(arquivo, "%d ", vetorOrdenado[i]);
            }

            fprintf(arquivo, "\n");

            switch (opcaosort) {
              case 1:
                fprintf(arquivo, "Método de ordenação usado: Merge Sort\n");
                break;

              case 2:
                fprintf(arquivo, "Método de ordenação usado: Bubble Sort\n");
                break;

            case 3:
              fprintf(arquivo, "Método de ordenação usado: Heap Sort\n");
              break;

            case 4:
              fprintf(arquivo, "Método de ordenação usado: Outro método\n");
              break;
          }

            fprintf(arquivo, "Quantidade de Execuções: %d", QTD);

            fclose(arquivo);

             printf("=============================================\n");
             break;
      }
        case 8: {
          FILE *arquivo;
          int quantidade = 0;
          int numero;

          while(1){
            printf("Escolha um dos arquivos a seguir para ser importado:\n");
            printf("1- Arquivo 1: Dados aleatórios e tamanho 100\n");
            printf("2- Arquivo 2: Dados aleatórios e tamanho 500\n");
            printf("3- Arquivo 3: Dados aleatórios e tamanho 1K\n");
            scanf("%d", &numero);

            switch (numero) {
              case 1:
                arquivo = fopen("/content/dtaleat1kuni2.txt", "r");
                break;
              case 2:
                arquivo = fopen("/content/dtaleat1kuni2.txt", "r");
                break;
              case 3:
                arquivo = fopen("/content/dtaleat1kuni2.txt", "r");
                break;
              default:
                printf("Opção inválida!\n");
                continue;
            }

            if (arquivo == NULL) {
              printf("Erro ao abrir o arquivo.\n");
              break;
            }
          while (fscanf(arquivo, "%d", &numero) == 1) {
            quantidade++;
          }
          fclose(arquivo);
          if (quantidade == 0) {
            printf("O arquivo está vazio.\n");
            fclose(arquivo);
            break;
          }
        }

    // Volta para o começo
      rewind(arquivo);

    // Libera os vetores antigos
      free(vetorOriginal);
      free(vetorOrdenado);

    // Cria os novos vetores
      vetorOriginal = malloc(quantidade * sizeof(int));
      vetorOrdenado = malloc(quantidade * sizeof(int));

      if (vetorOriginal == NULL || vetorOrdenado == NULL) {
        printf("Erro de alocacao de memoria!\n");
        fclose(arquivo);
        break;
      }

    // Coloca os números nos vetores
      for (int i = 0; i < quantidade; i++) {
        fscanf(arquivo, "%d", &vetorOriginal[i]);

        vetorOrdenado[i] = vetorOriginal[i];
      }

    fclose(arquivo);

    tamanho_usuario = quantidade;

    printf("Arquivo importado com sucesso!\n");
    printf("Quantidade de numeros: %d\n", quantidade);

    imp_done = 1;

    break;
  }


  case 9: {
    free(vetorOriginal);
    free(vetorOrdenado);

    vetorOriginal = NULL;
    vetorOrdenado = NULL;

    printf("Saindo...\n");
    return 0;
  }
  default: // números maiores que 8 não são válidos
      printf("Opção inválida!\n");
      continue;
    }
  }
}
```

```python
%%writefile heapSort.c

// Heap Sort
#include <stdio.h>
#include <stdlib.h>
#include <time.h>

#define TAM(x) (sizeof(x) / sizeof((x)[0])) // Função para verificar o tamanho do vetor

long long count = 0;
#define SIMAS(x) (count += (x))


int ContaLinhas_GuardaV(int **vetor)
{
    FILE *arquivo;

    int capacidade = 0;

    arquivo = fopen("/content/drive/MyDrive/Coisas da escola/APS_SIMAS/TXTs/ComduplicidadeAleatorio_1K.txt", "r");

    if (arquivo == NULL)
    {
        printf("Erro ao abrir o arquivo.\n");
        return 1;
    }

    // Conta a quantidade de linhas
    int numero;

    while (fscanf(arquivo, "%d", &numero) == 1)
    {
        capacidade++;
    }

    printf("Quantidade de linhas: %d\n", capacidade);
    printf("\n");

    // Volta para o início do arquivo
    rewind(arquivo);

    // Cria o vetor com exatamente a quantidade necessária
    *vetor = malloc(capacidade * sizeof(int));

    if (*vetor == NULL)
    {
        printf("Erro ao alocar memoria.\n");
        fclose(arquivo);
        return 1;
    }

    // Preenche o vetor
    for (int i = 0; i < capacidade; i++)
    {
        fscanf(arquivo, "%d", &(*vetor)[i]);
    }

    fclose(arquivo);

    return capacidade;
}


void criaHeap(int *vet, int i, int f)
{
    int aux = vet[i];
    SIMAS(1);

    int j = i * 2 + 1; // Primeiro filho
    SIMAS(3);

    while (j <= f)
    {
        SIMAS(1);

        SIMAS(1);
        if(j < f)
        {
            SIMAS(2);

            if(vet[j] < vet[j + 1])
            {
                // Verifica o segundo filho
                j = j + 1;
                SIMAS(1);
            }
        }

        SIMAS(1);

        if(aux < vet[j])
        {
            vet[i] = vet[j];
            SIMAS(1);

            i = j;
            SIMAS(1);

            j = 2 * i + 1;
            SIMAS(1);
        }
        else
        {
            j = f + 1;
            SIMAS(1);
        }
    }

    vet[i] = aux;
    SIMAS(1);
}


void heapSort(int *vet, int N)
{
    int i, aux;

    SIMAS(3);

    for(i = (N - 1) / 2; i >= 0; i--)
    {
        criaHeap(vet, i, N - 1);
        SIMAS(1);
    }

    SIMAS(3);

    for (i = N - 1; i >= 1; i--)
    {
        aux = vet[0];
        SIMAS(1);

        vet[0] = vet[i];
        SIMAS(1);

        vet[i] = aux;
        SIMAS(1);

        criaHeap(vet, 0, i - 1);
        SIMAS(1);
    }
}


int main(void)
{
    clock_t start = clock();

    // Vetor
    int *vetor;

    int tamanho = ContaLinhas_GuardaV(&vetor);

    // Vetor desordenado
    printf("Vetor desordenado: ");

    for (int i = 0; i < tamanho; i++)
    {
        printf("%d | ", vetor[i]);
    }

    printf("\n \n");

    // Uso da função Heap Sort
    heapSort(vetor, tamanho);

    // Vetor ordenado
    printf("Vetor ordenado: ");

    for (int i = 0; i < tamanho; i++)
    {
        printf("%d | ", vetor[i]);
    }

    printf("\n");

    // Simulação de carga de processamento
    for (volatile int i = 0; i < 100000000; i++)
    {
    }

    // Registra o tempo final
    clock_t final = clock();

    // Calcula a diferença e converte para segundos
    double process_time = (double)(final - start) / CLOCKS_PER_SEC;

    printf("\nTempo de compilacao: %f segundos\n", process_time);

    printf("\nContagem de execuções: %lld vezes\n", count);

    free(vetor);

    return 0;
}
```
